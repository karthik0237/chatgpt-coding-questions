```python
from collections import Counter

logs = [
    "2026-11-13 08:00:00 INFO Pipeline started",
    "2026-11-13 08:01:00 ERROR Database connection failed",
    "2026-11-13 08:02:00 INFO Reading source data",
    "2026-11-13 08:03:00 WARN Missing values detected",
    "2026-11-13 08:04:00 ERROR Pipeline failed"
]
counts = Counter()
for log in logs:
    parts = log.split(maxsplit = 3) # or log.split(" ",3) splits string into 4 parts based on spaces
    if len(parts) >= 3:
        counts[parts[2]] +=1
print(dict(counts))
```

```python
'''
split(maxsplit=3)
separates date, time, log level and remaining message.Counter automatically maintains frequencies.
Complexity: O(n), where n is the number of log lines.
'''
```
