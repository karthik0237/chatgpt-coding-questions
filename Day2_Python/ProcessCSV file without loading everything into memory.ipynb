```python
# Process a large CSV file without loading everything into memory
# Calculate total sales amount for each region.
import csv
from decimal import Decimal

totals = {}
with open("/Volumes/chatcat/chatschema/chatvolume/day2Q3data.csv","r",newline = "") as file:
    reader = csv.DictReader(file)
    for row in reader:
        region = row["region"]
        amount = Decimal(row["amount"])
        totals[region] = totals.get(region, Decimal("0")) + amount
print(totals)



    
```
