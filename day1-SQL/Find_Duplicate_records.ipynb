```python
%sql
/* find duplicate records */
create catalog if not exists chatcat;

create schema if not exists chatcat.chatschema;

use chatcat.chatschema;

show tables;

select * from chatschema.customer1;
```

```python
%sql
use chatschema;
CREATE OR REPLACE TABLE chatschema.customer1(
    customer_id INT,
    name VARCHAR(50) NOT NULL,
    email VARCHAR(50)
);
```

```python
%sql
use chatcat.chatschema;

INSERT INTO TABLE chatschema.customer1(
    customer_id,name,email) 
    VALUES(1,"karthik","karthik@gmail.com"),
    (2,"ravi","ravi@gmail.com"),
    (3,"kiran","kiran@gmail.com"),
    (4,"ravi","ravi@gmail.com"),
    (5,"karthik","karthik@gmail.com");
```

```python
%sql
/* Write an SQL query to find duplicate email addresses. */
USE chatcat.chatschema;

SELECT email, COUNT(*) AS cnt
FROM chatschema.customer1
GROUP BY email
HAVING COUNT(*) > 1;
```
