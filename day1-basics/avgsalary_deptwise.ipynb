```python
from pyspark.sql import functions as F
data = [
    (1, "Ravi", "IT", 60000),
    (2, "Kumar", "IT", 80000),
    (3, "Priya", "HR", 40000),
    (4, "Arun", "HR", 60000)
]

columns = ["id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)
df.show()

avgdf = df.groupBy("department")\
        .agg(
            F.avg("salary").alias("avg_salary")
            )
avgdf.show()
```
