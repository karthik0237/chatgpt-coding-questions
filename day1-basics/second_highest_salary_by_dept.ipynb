```python
%sql
USE chatcat.chatschema;

CREATE TABLE chatschema.employees (
    emp_id INT PRIMARY KEY,
    name VARCHAR(100),
    department VARCHAR(50),
    salary INT
);
```

```python
%sql
/* Write a query to find the second-highest distinct salary in each department */

WITH ranktab AS (
    SELECT 
    department, 
    salary, 
    DENSE_RANK() OVER(
        PARTITION BY department
        ORDER BY salary DESC
    ) AS rnk
    FROM employees
)

SELECT department, salary FROM ranktab WHERE rnk = 2;

```
