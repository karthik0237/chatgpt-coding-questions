```python
# parse timestamps and extract dates

from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

# Define schema
schema = StructType([
    StructField("event_id", IntegerType(), True),
    StructField("user_id", IntegerType(), True),
    StructField("event_timestamp", StringType(), True)
])

# Sample data
data = [
    (1, 101, "2026-11-14 08:30:00"),
    (2, 102, "2026-11-14 09:15:00"),
    (3, 103, "2026-11-13 18:45:00"),
    (4, 104, "invalid_date")
]

# Create DataFrame
df = spark.createDataFrame(data, schema)

# Display DataFrame
df.show(truncate=False)

# Print schema
df.printSchema()
```

```python
from pyspark.sql import functions as F
parsed_df = df.withColumn("event_ts", F.try_to_timestamp("event_timestamp", F.lit("yyyy-MM-dd HH:mm:ss")))
parsed_df.show()
```

```python
valid_df = parsed_df.filter(F.col("event_ts").isNotNull())
invalid_df  = parsed_df.filter(F.col("event_ts").isNull())
valid_df.show()
invalid_df.show()
```
